## 9. Мини-проект: Угадай число
**Задание:**
1. Используйте модуль `random`, чтобы загадать секретное число от 1 до 10 (`import random`, `secret_number = random.randint(1, 10)`).
2. Попросите пользователя угадать число.
3. Используйте цикл `while`, чтобы продолжать спрашивать, пока пользователь не угадает правильно.
4. Если его догадка слишком велика, выведите "Слишком много!". Если слишком мала - "Слишком мало!".
5. Когда он угадает правильно, выведите поздравительное сообщение.

```python
import random
# Ваш код здесь
secret_number = random.randint(1,10)
user_number = -1
while(user_number != secret_number):
    user_number = int(input('Think a number'))
    if(user_number < secret_number):
        print("Слишком мало!")
    elif(user_number > secret_number):
         print("Слишком много!")
    
    if(user_number == secret_number):
        print("Congratulations!")

```
