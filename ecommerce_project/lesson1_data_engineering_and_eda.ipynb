### Задание 1:
Напишите SQL-запрос, чтобы объединить все три таблицы (JOIN) и получить единый датасет со следующими колонками:
`transaction_id`, `date`, `customer_id`, `age`, `gender`, `location`, `product_id`, `category`, `price`, `quantity`.

Сохраните результат в датафрейм `df`.

```python
query = """
SELECT 
    t.transaction_id,
    t.date,
    c.customer_id,
    c.age,
    c.gender,
    c.location,
    p.product_id,
    p.category,
    p.price,
    t.quantity
FROM transactions t
JOIN customers c ON t.customer_id = c.customer_id
JOIN products p ON t.product_id = p.product_id
"""

# Выгружаем данные
df = pd.read_sql(query, conn)

# Закрываем соединение с БД
conn.close()

df.head()
```
