# Тема 8: Очистка данных (ОТВЕТЫ)

```python
import pandas as pd
import numpy as np
df = pd.merge(pd.read_csv('data/transactions.csv'), pd.read_csv('data/customers.csv'), on='customer_id')
df = pd.merge(df, pd.read_csv('data/products.csv'), on='product_id')
df['total_sum'] = df['price'] * df['quantity']
```

### Задание 1

```python
df.isnull().sum()
```

### Задание 2

```python
pct = df['age'].isnull().sum() / len(df) * 100
print(f"Процент пропусков в age: {pct:.2f}%")
```

### Задание 3

```python
df['age'] = df['age'].fillna(df['age'].median())
print(f"Пропусков: {df['age'].isnull().sum()}")
```

### Задание 4

```python
print(f"Дубликатов: {df.duplicated().sum()}")
df = df.drop_duplicates()
print(f"После удаления: {df.shape}")
```

### Задание 5

```python
df['date'] = pd.to_datetime(df['date'])
print(df['date'].dtype)
```

### Задание 6

```python
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day_of_week'] = df['date'].dt.dayofweek
df[['date', 'year', 'month', 'day_of_week']].head()
```

### Задание 7

```python
Q1 = df['total_sum'].quantile(0.25)
Q3 = df['total_sum'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df['total_sum'] < lower) | (df['total_sum'] > upper)]
print(f"Выбросов: {len(outliers)}")
```

### Задание 8

```python
df['total_sum'] = df['total_sum'].clip(lower=lower, upper=upper)
print(df['total_sum'].describe())
```

### Задание 9

```python
bad_rows = df[(df['quantity'] <= 0) | (df['price'] <= 0)]
print(f"Логических ошибок: {len(bad_rows)}")
df = df[(df['quantity'] > 0) & (df['price'] > 0)]
```

### Задание 10

```python
df['location'] = df['location'].str.strip().str.title()
print(df['location'].unique())
```

### Задание 11

```python
print('Размер:', df.shape)
print('\nПропуски:\n', df.isnull().sum())
print('\nТипы:\n', df.dtypes)
```

### Задание 12

```python
df.to_csv('data/cleaned_data.csv', index=False)
print('Файл сохранён!')
```
