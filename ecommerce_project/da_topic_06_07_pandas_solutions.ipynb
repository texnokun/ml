# Темы 6-7: Библиотека Pandas (ОТВЕТЫ)

```python
import pandas as pd
customers = pd.read_csv('data/customers.csv')
products = pd.read_csv('data/products.csv')
transactions = pd.read_csv('data/transactions.csv')
```

### Задание 1

```python
print(customers.head(10))
print(customers.dtypes)
```

```python
transactions.describe()
```

### Задание 3

```python
moscow = customers[customers['location'] == 'Moscow']
print(f"Клиентов из Москвы: {len(moscow)}")
```

### Задание 4

```python
df = pd.merge(transactions, customers, on='customer_id')
df = pd.merge(df, products, on='product_id')
print(df.shape)
df.head()
```

### Задание 5

```python
df['total_sum'] = df['price'] * df['quantity']
df.nlargest(5, 'total_sum')[['transaction_id', 'category', 'price', 'quantity', 'total_sum']]
```

### Задание 6

```python
df.groupby('category')['total_sum'].sum().sort_values(ascending=False)
```

### Задание 7

```python
df.groupby('location')['total_sum'].mean().sort_values(ascending=False)
```

### Задание 8

```python
df.groupby('category')['total_sum'].agg(['count', 'mean', 'max'])
```

### Задание 9

```python
top10 = df.groupby(['customer_id', 'location'])['total_sum'].sum().reset_index()
top10.nlargest(10, 'total_sum')
```

### Задание 10

```python
df['date'] = pd.to_datetime(df['date'])
df['month'] = df['date'].dt.to_period('M')
df.groupby('month')['total_sum'].sum()
```
