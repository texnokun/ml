## Что такое машинное обучение?
Машинное обучение — это способность компьютера находить закономерности в данных **без явного программирования правил**.

### Пример из нашего проекта:
- **Без ML:** Аналитик вручную смотрит на графики и говорит: «Кажется, клиенты старше 40 тратят больше».
- **С ML:** Алгоритм сам находит, что комбинация (возраст > 38, город = Москва, покупал Electronics) → с вероятностью 73% потратит > 5000 руб.

## Виды машинного обучения

| Тип | Описание | Наш пример |
|---|---|---|
| **Supervised (с учителем)** | Есть целевая переменная (y) | Предсказать сумму покупки, предсказать отток |
| **Unsupervised (без учителя)** | Нет целевой переменной | Сегментация клиентов на группы |

### Supervised Learning делится на:
- **Регрессия** — предсказываем число (сколько потратит клиент?)
- **Классификация** — предсказываем категорию (уйдёт клиент или нет?)
