# Тема 14: Основы статистики (ОТВЕТЫ)

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.merge(pd.read_csv('data/transactions.csv'), pd.read_csv('data/customers.csv'), on='customer_id')
df = pd.merge(df, pd.read_csv('data/products.csv'), on='product_id')
df['total_sum'] = df['price'] * df['quantity']
df['age'] = df['age'].fillna(df['age'].median())
```

### Задание 1

```python
print(f"Среднее: {df['total_sum'].mean():.2f}")
print(f"Медиана: {df['total_sum'].median():.2f}")
print(f"Мода: {df['total_sum'].mode()[0]:.2f}")
# Медиана лучше, так как данные содержат выбросы (асимметричное распределение)
```

### Задание 2

```python
print(f"Стд. отклонение: {df['total_sum'].std():.2f}")
print(f"Дисперсия: {df['total_sum'].var():.2f}")
print(f"Размах: {df['total_sum'].max() - df['total_sum'].min():.2f}")
```

### Задание 3

```python
for p in [25, 50, 75, 90]:
    print(f"{p}-й перцентиль: {df['total_sum'].quantile(p/100):.2f}")
# 90-й перцентиль: 90% покупок ниже этого значения
```

### Задание 4

```python
corr = df['age'].corr(df['total_sum'])
print(f"Корреляция age-total_sum: {corr:.4f}")
plt.figure(figsize=(8,5))
sns.scatterplot(data=df.sample(500), x='age', y='total_sum', alpha=0.5)
plt.title(f'Корреляция: {corr:.4f}')
plt.show()
```

### Задание 5

```python
male = df[df['gender']=='M']['total_sum']
female = df[df['gender']=='F']['total_sum']
t_stat, p_value = stats.ttest_ind(male, female, equal_var=False)
print(f"T-stat: {t_stat:.4f}, P-value: {p_value:.4f}")
print('Различия значимы!' if p_value < 0.05 else 'Различия НЕ значимы.')
```

### Задание 6

```python
df['date'] = pd.to_datetime(df['date'])
df['is_weekend'] = df['date'].dt.dayofweek >= 5
weekend = df[df['is_weekend']]['total_sum']
weekday = df[~df['is_weekend']]['total_sum']
t_stat, p_value = stats.ttest_ind(weekend, weekday, equal_var=False)
print(f"T-stat: {t_stat:.4f}, P-value: {p_value:.4f}")
print('Выходные отличаются!' if p_value < 0.05 else 'Разницы нет.')
```

### Задание 7

```python
ct = pd.crosstab(df['gender'], df['category'])
chi2, p, dof, expected = stats.chi2_contingency(ct)
print(f"Chi2: {chi2:.4f}, P-value: {p:.4f}")
print('Зависимость есть!' if p < 0.05 else 'Независимы.')
```

### Задание 8

```python
sample = df['total_sum'].sample(500, random_state=42)
stat, p = stats.shapiro(sample)
print(f"Shapiro p-value: {p:.6f}")
plt.figure(figsize=(8,5))
sns.histplot(df['total_sum'], bins=50, stat='density')
plt.title('Распределение total_sum')
plt.show()
```

### Задание 9

```python
mean = df['total_sum'].mean()
std = df['total_sum'].std()
n = len(df)
ci_low = mean - 1.96 * (std / np.sqrt(n))
ci_high = mean + 1.96 * (std / np.sqrt(n))
print(f"95% ДИ: [{ci_low:.2f}, {ci_high:.2f}]")
```

### Задание 10

```python
cities = df['location'].unique()
groups = [df[df['location']==c]['total_sum'].values for c in cities]
f_stat, p_value = stats.f_oneway(*groups)
print(f"F-stat: {f_stat:.4f}, P-value: {p_value:.4f}")
print('Средние различаются!' if p_value < 0.05 else 'Средние НЕ различаются.')
```
