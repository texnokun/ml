### Задание 9: Зачем нужен Test Set?

**Ответ:**

Если обучить модель на всех данных, она «запомнит» ответы (как школьник, который выучил ответы на тест, а не понял тему). 
Это называется **переобучение (overfitting)**. Модель будет отлично работать на известных данных, 
но провалится на новых, реальных клиентах.

Тестовая выборка (Test Set) — это «экзамен», который модель видит впервые. 
Если она показывает хорошие результаты на тесте — значит, она действительно научилась, а не зазубрила.

### Задание 10: Словарь терминов ML

| Термин | Определение |
|---|---|
| Feature (Признак) | Входная переменная, на основе которой модель делает предсказание (age, price, quantity) |
| Target (Целевая переменная) | То, что мы хотим предсказать (total_sum, churn) |
| Training Set | Данные, на которых модель учится (обычно 80% от всех данных) |
| Test Set | Данные для проверки качества модели (обычно 20%, модель их НЕ видела при обучении) |
| Overfitting (Переобучение) | Модель «зазубрила» обучающие данные, но плохо работает на новых |
| Underfitting (Недообучение) | Модель слишком простая и не улавливает закономерности в данных |
