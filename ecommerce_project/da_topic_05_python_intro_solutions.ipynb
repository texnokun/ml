### Задание 2

```python
daily_revenue = [45000, 52000, 38000, 61000, 73000, 89000, 41000]
total = sum(daily_revenue)
average = total / len(daily_revenue)
print(f"Общая выручка: {total}")
print(f"Средний доход: {average:.0f}")
print(f"Макс: {max(daily_revenue)}, Мин: {min(daily_revenue)}")
```

### Задание 4

```python
products = [
    {'name': 'Ноутбук', 'price': 45000, 'category': 'Electronics'},
    {'name': 'Футболка', 'price': 1200, 'category': 'Clothing'},
    {'name': 'Книга Python', 'price': 890, 'category': 'Books'},
    {'name': 'Кроссовки', 'price': 5600, 'category': 'Sports'},
    {'name': 'Лампа', 'price': 2300, 'category': 'Home & Garden'},
]
for p in products:
    print(f"{p['name']} — {p['price']} руб.")
```

### Задание 5

```python
expensive = []
for p in products:
    if p['price'] > 2000:
        expensive.append(p)
print(expensive)
```

### Задание 6

```python
def apply_discount(price, discount_percent):
    return price * (1 - discount_percent / 100)

for p in products:
    new_price = apply_discount(p['price'], 15)
    print(f"{p['name']}: {p['price']} → {new_price:.0f} руб.")
```

### Задание 7

```python
purchases = ['Electronics', 'Clothing', 'Books', 'Electronics', 'Sports',
             'Clothing', 'Electronics', 'Books', 'Clothing', 'Electronics',
             'Home & Garden', 'Electronics', 'Books', 'Sports', 'Clothing']
category_count = {}
for cat in purchases:
    if cat in category_count:
        category_count[cat] += 1
    else:
        category_count[cat] = 1
print(category_count)
```

### Задание 8

```python
cart = [
    {'product': 'Наушники', 'price': 3500, 'quantity': 1},
    {'product': 'Чехол для телефона', 'price': 800, 'quantity': 2},
    {'product': 'Зарядное устройство', 'price': 1200, 'quantity': 1},
]
total = sum(item['price'] * item['quantity'] for item in cart)
print(f"Итого: {total} руб.")
```

### Задание 10

```python
def generate_report(transactions):
    return {
        'total': sum(transactions),
        'count': len(transactions),
        'average': sum(transactions) / len(transactions),
        'max': max(transactions),
        'min': min(transactions)
    }

transactions = [1500, 3200, 450, 990, 12000, 780, 2340, 5600, 870, 4100]
print(generate_report(transactions))
```

### Задание 11

```python
raw_csv = """customer_id,name,age
101,Иван,34
102,Мария,28
103,Алексей,45"""

lines = raw_csv.strip().split('\n')
headers = lines[0].split(',')
result = []
for line in lines[1:]:
    values = line.split(',')
    result.append(dict(zip(headers, values)))
print(result)
```

### Задание 12

```python
clients = [
    {'name': 'Иван', 'total_spent': 15000},
    {'name': 'Мария', 'total_spent': 42000},
    {'name': 'Алексей', 'total_spent': 8700},
    {'name': 'Елена', 'total_spent': 31000},
    {'name': 'Дмитрий', 'total_spent': 27500},
]
top3 = sorted(clients, key=lambda x: x['total_spent'], reverse=True)[:3]
for i, c in enumerate(top3, 1):
    print(f"{i}. {c['name']} — {c['total_spent']} руб.")
```
