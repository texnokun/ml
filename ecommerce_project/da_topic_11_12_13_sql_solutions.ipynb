### Задание 5

```python
pd.read_sql("""
SELECT t.transaction_id, c.customer_id, c.location, p.category, p.price, t.quantity,
       p.price * t.quantity AS revenue
FROM transactions t
JOIN customers c ON t.customer_id = c.customer_id
JOIN products p ON t.product_id = p.product_id
LIMIT 20
""", conn)
```

### Задание 6

```python
pd.read_sql("""
SELECT p.category, SUM(p.price * t.quantity) AS total_revenue
FROM transactions t
JOIN products p ON t.product_id = p.product_id
GROUP BY p.category
ORDER BY total_revenue DESC
""", conn)
```

### Задание 7

```python
pd.read_sql("""
SELECT c.location, SUM(p.price * t.quantity) AS total_revenue
FROM transactions t
JOIN customers c ON t.customer_id = c.customer_id
JOIN products p ON t.product_id = p.product_id
GROUP BY c.location
HAVING total_revenue > 100000
ORDER BY total_revenue DESC
""", conn)
```

### Задание 8

```python
pd.read_sql("""
SELECT c.customer_id, c.location, SUM(p.price * t.quantity) AS total_spent
FROM transactions t
JOIN customers c ON t.customer_id = c.customer_id
JOIN products p ON t.product_id = p.product_id
GROUP BY c.customer_id, c.location
ORDER BY total_spent DESC
LIMIT 5
""", conn)
```

### Задание 9

```python
pd.read_sql("""
SELECT customer_id, AVG(price * quantity) AS avg_check
FROM (
    SELECT t.customer_id, p.price, t.quantity
    FROM transactions t
    JOIN products p ON t.product_id = p.product_id
)
GROUP BY customer_id
HAVING avg_check > (
    SELECT AVG(p.price * t.quantity)
    FROM transactions t
    JOIN products p ON t.product_id = p.product_id
)
ORDER BY avg_check DESC
""", conn)
```

### Задание 10

```python
pd.read_sql("""
SELECT
    CASE
        WHEN age < 25 THEN 'Молодой'
        WHEN age BETWEEN 25 AND 45 THEN 'Средний'
        ELSE 'Старший'
    END AS age_group,
    COUNT(*) AS cnt
FROM customers
WHERE age IS NOT NULL
GROUP BY age_group
""", conn)
```

### Задание 11

```python
pd.read_sql("""
SELECT strftime('%Y-%m', t.date) AS month,
       AVG(p.price * t.quantity) AS avg_check
FROM transactions t
JOIN products p ON t.product_id = p.product_id
GROUP BY month
ORDER BY month
""", conn)
```

### Задание 12

```python
pd.read_sql("""
SELECT customer_id, total_spent,
       ROW_NUMBER() OVER (ORDER BY total_spent DESC) AS rank
FROM (
    SELECT c.customer_id, SUM(p.price * t.quantity) AS total_spent
    FROM transactions t
    JOIN customers c ON t.customer_id = c.customer_id
    JOIN products p ON t.product_id = p.product_id
    GROUP BY c.customer_id
)
LIMIT 10
""", conn)
```
