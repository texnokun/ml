# Темы 9-10: Визуализация данных (ОТВЕТЫ)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.merge(pd.read_csv('data/transactions.csv'), pd.read_csv('data/customers.csv'), on='customer_id')
df = pd.merge(df, pd.read_csv('data/products.csv'), on='product_id')
df['total_sum'] = df['price'] * df['quantity']
df['date'] = pd.to_datetime(df['date'])
df['age'] = df['age'].fillna(df['age'].median())
```

### Задание 1

```python
plt.figure(figsize=(8,5))
sns.histplot(df['age'], bins=20, kde=True)
plt.title('Распределение возраста покупателей')
plt.xlabel('Возраст')
plt.ylabel('Количество')
plt.show()
```

### Задание 2

```python
rev = df.groupby('category')['total_sum'].sum().sort_values().reset_index()
plt.figure(figsize=(10,5))
sns.barplot(data=rev, x='total_sum', y='category')
plt.title('Выручка по категориям')
plt.show()
```

### Задание 3

```python
plt.figure(figsize=(12,5))
sns.boxplot(data=df, x='category', y='total_sum')
plt.title('Распределение чека по категориям')
plt.xticks(rotation=15)
plt.show()
```

### Задание 4

```python
monthly = df.groupby(df['date'].dt.to_period('M'))['total_sum'].sum().reset_index()
monthly['date'] = monthly['date'].astype(str)
plt.figure(figsize=(12,5))
plt.plot(monthly['date'], monthly['total_sum'], marker='o')
plt.title('Динамика выручки по месяцам')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Задание 5

```python
city_counts = df.groupby('location')['customer_id'].nunique()
plt.figure(figsize=(7,7))
plt.pie(city_counts, labels=city_counts.index, autopct='%1.1f%%')
plt.title('Доля клиентов по городам')
plt.show()
```

### Задание 6

```python
corr = df[['age','price','quantity','total_sum']].corr()
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Матрица корреляций')
plt.show()
```

### Задание 7

```python
df['day_of_week'] = df['date'].dt.dayofweek
plt.figure(figsize=(8,5))
sns.countplot(x='day_of_week', data=df)
plt.title('Покупки по дням недели (0=Пн)')
plt.show()
```

### Задание 8

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df.sample(500), x='age', y='total_sum', hue='gender', alpha=0.6)
plt.title('Возраст vs Сумма покупки')
plt.show()
```

### Задание 9

```python
fig, axes = plt.subplots(1, 2, figsize=(14,5))
sns.histplot(df['total_sum'], bins=30, ax=axes[0])
axes[0].set_title('Распределение суммы чека')
sns.boxplot(data=df, x='gender', y='total_sum', ax=axes[1])
axes[1].set_title('Чек по полу')
plt.tight_layout()
plt.show()
```

### Задание 10

```python
plt.figure(figsize=(8,5))
sns.violinplot(data=df, x='gender', y='total_sum')
plt.title('Violin plot: чек по полу')
# Violin plot показывает плотность распределения, а не только квартили
plt.show()
```

### Задание 11

```python
subset = df[['age','price','quantity','total_sum','gender']].sample(300)
sns.pairplot(subset, hue='gender')
plt.show()
```

### Задание 12

```python
sns.set_style('whitegrid')
sns.set_context('talk')
plt.figure(figsize=(10,6))
rev = df.groupby('category')['total_sum'].sum().sort_values().reset_index()
sns.barplot(data=rev, x='total_sum', y='category', palette='viridis')
plt.title('Выручка по категориям товаров', fontsize=18)
plt.xlabel('Выручка (руб.)')
plt.ylabel('')
plt.tight_layout()
plt.savefig('report_chart.png', dpi=150)
plt.show()
```
