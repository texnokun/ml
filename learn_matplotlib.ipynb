# Руководство по Matplotlib
В этом ноутбуке рассматриваются основы создания визуализаций с использованием `matplotlib`.

## Основные методы NumPy (`np`) и Matplotlib (`plt`)

**NumPy (`np`)** — библиотека для работы с многомерными массивами и математическими функциями.
- `np.linspace(start, stop, num)`: Возвращает `num` равномерно распределенных чисел на интервале `[start, stop]`.
- `np.sin(x)`: Вычисляет тригонометрический синус для каждого элемента массива `x`.
- `np.random.rand(d0, d1, ...)`: Генерирует случайные значения из равномерного распределения на интервале `[0, 1)`.
- `np.random.seed(seed)`: Инициализирует генератор случайных чисел для воспроизводимости результатов.
- `np.random.randn(d0, d1, ...)`: Генерирует случайные значения из стандартного нормального распределения (среднее 0, дисперсия 1).

**Matplotlib Pyplot (`plt`)** — интерфейс для создания графиков.
- `plt.figure(figsize=(w, h))`: Создает новую фигуру (окно для графика) заданного размера (ширина `w`, высота `h` в дюймах).
- `plt.plot(x, y, ...)`: Строит линейный график по точкам `x` и `y`.
- `plt.scatter(x, y, ...)`: Строит диаграмму рассеяния (точечный график).
- `plt.bar(x, height, ...)`: Строит столбчатую диаграмму.
- `plt.hist(x, bins=...)`: Строит гистограмму (распределение частот).
- `plt.subplots(nrows, ncols, ...)`: Создает фигуру и набор подграфиков (осей).
- `plt.title('...')`: Устанавливает заголовок для текущего графика.
- `plt.xlabel('...')`, `plt.ylabel('...')`: Устанавливают подписи для осей X и Y соответственно.
- `plt.legend()`: Отображает легенду на графике (используя метки `label` из функций построения).
- `plt.grid(True)`: Отображает сетку на фоне графика.
- `plt.tight_layout()`: Автоматически подстраивает параметры подграфиков, чтобы они помещались в фигуру без наложений.
- `plt.show()`: Отображает все открытые фигуры (обязательно в конце для отрисовки графиков).

```python
import matplotlib.pyplot as plt # Импорт модуля pyplot из библиотеки matplotlib под псевдонимом plt
import numpy as np              # Импорт библиотеки numpy под псевдонимом np

%matplotlib inline             
# Магическая команда Jupyter для отображения графиков прямо в ноутбуке
```

## Простой линейный график
Построим простой график функции синуса.

```python
x = np.linspace(0, 10, 100)                      # Создание массива из 100 равномерно распределенных точек от 0 до 10
y = np.sin(x)                                    # Вычисление значения синуса для каждой точки из массива x

plt.figure(figsize=(8, 4))                       # Создание новой фигуры графика размером 8 на 4 дюйма
plt.plot(x, y, label='sin(x)', color='blue')     # Построение линейного графика по точкам x и y синим цветом с меткой 'sin(x)'
plt.title('Простой линейный график: Синусоида')  # Установка заголовка графика
plt.xlabel('Ось X')                              # Установка подписи для оси X
plt.ylabel('Ось Y')                              # Установка подписи для оси Y
plt.legend()                                     # Отображение легенды (в данном случае для 'sin(x)')
plt.grid(True)                                   # Включение отображения сетки на фоне графика
plt.show()                                       # Отрисовка и отображение готового графика
```

## Диаграмма рассеяния (Scatter Plot)
Диаграммы рассеяния отлично подходят для отображения зависимости между двумя переменными.

```python
np.random.seed(42)                                                 # Инициализация генератора случайных чисел для воспроизводимости
x = np.random.rand(50)                                             # Генерация массива из 50 случайных чисел от 0 до 1 для оси X
y = np.random.rand(50)                                             # Генерация массива из 50 случайных чисел от 0 до 1 для оси Y
colors = np.random.rand(50)                                        # Генерация массива из 50 случайных чисел для определения цвета каждой точки
area = (30 * np.random.rand(50))**2                                # Расчет площади для каждой точки (радиус от 0 до 30)

plt.figure(figsize=(8, 5))                                         # Создание новой фигуры графика размером 8 на 5 дюймов
plt.scatter(x, y, s=area, c=colors, alpha=0.5, cmap='viridis')     # Построение точечного графика с заданными координатами, размерами (s), цветами (c), прозрачностью (alpha) и цветовой картой
plt.title('Пример диаграммы рассеяния')                            # Установка заголовка графика
plt.colorbar()                                                     # Отображение цветовой шкалы справа от графика
plt.show()                                                         # Отрисовка и отображение готового графика
```

## Столбчатая диаграмма (Bar Chart)
Столбчатые диаграммы полезны для сравнения категориальных данных.

```python
categories = ['Категория A', 'Категория B', 'Категория C', 'Категория D']               # Список названий категорий
values = [15, 24, 12, 32]                                                               # Список значений, соответствующих каждой категории

plt.figure(figsize=(8, 4))                                                              # Создание новой фигуры графика размером 8 на 4 дюйма
plt.bar(categories, values, color=['skyblue', 'lightgreen', 'salmon', 'orchid'])        # Построение столбчатой диаграммы с заданными категориями, значениями и цветами столбцов
plt.title('Пример столбчатой диаграммы')                                                # Установка заголовка графика
plt.ylabel('Значения')                                                                  # Установка подписи для оси Y
plt.show()                                                                              # Отрисовка и отображение готового графика
```

## Гистограмма
Гистограммы показывают распределение набора данных.

```python
data = np.random.randn(1000)                                                  # Генерация массива из 1000 чисел со стандартным нормальным распределением

plt.figure(figsize=(8, 4))                                                    # Создание новой фигуры графика размером 8 на 4 дюйма
plt.hist(data, bins=30, color='purple', alpha=0.7, edgecolor='black')         # Построение гистограммы данных, разбитых на 30 корзин (bins), с заданной прозрачностью и обводкой
plt.title('Пример гистограммы')                                               # Установка заголовка графика
plt.xlabel('Значение')                                                        # Установка подписи для оси X
plt.ylabel('Частота')                                                         # Установка подписи для оси Y
plt.show()                                                                    # Отрисовка и отображение готового графика
```

## Несколько графиков (Subplots)
Вы можете отображать несколько графиков в одной фигуре, используя `subplots`.

```python
fig, axs = plt.subplots(2, 2, figsize=(10, 8))                        # Создание фигуры (fig) и сетки осей (axs) размером 2 на 2 подграфика, общий размер фигуры 10x8 дюймов

# Верхний левый график (индекс [0, 0])
axs[0, 0].plot(x, y, 'tab:blue')                                       # Построение линейного графика синего цвета ('tab:blue')
axs[0, 0].set_title('График 1: Линейный')                              # Установка заголовка для верхнего левого графика

# Верхний правый график (индекс [0, 1])
axs[0, 1].scatter(x, y, c='tab:orange')                                # Построение точечного графика оранжевого цвета ('tab:orange')
axs[0, 1].set_title('График 2: Рассеяние')                             # Установка заголовка для верхнего правого графика

# Нижний левый график (индекс [1, 0])
axs[1, 0].bar(['A', 'B', 'C'], [3, 1, 2], color='tab:green')           # Построение столбчатой диаграммы зеленого цвета ('tab:green')
axs[1, 0].set_title('График 3: Столбчатый')                            # Установка заголовка для нижнего левого графика

# Нижний правый график (индекс [1, 1])
axs[1, 1].hist(np.random.randn(100), bins=20, color='tab:red')         # Построение гистограммы красного цвета ('tab:red') из 100 случайных чисел
axs[1, 1].set_title('График 4: Гистограмма')                           # Установка заголовка для нижнего правого графика

plt.tight_layout()                                                     # Автоматическая настройка отступов между подграфиками, чтобы избежать наложений
plt.show()                                                             # Отрисовка и отображение всей фигуры с подграфиками
```
