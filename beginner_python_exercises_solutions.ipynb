## 2. Базовая математика
**Задание:** Создайте программу, которая запрашивает у пользователя два числа. Вычислите и выведите их сумму, разность, произведение и частное.

*Подсказка:* Используйте функцию `input()` и не забудьте преобразовать введенные строки в числа с помощью `int()` или `float()`.

```python
num1 = float(input("Введите первое число: "))
num2 = float(input("Введите второе число: "))

print("Сумма:", num1 + num2)
print("Разность:", num1 - num2)
print("Произведение:", num1 * num2)
print("Частное:", num1 / num2)
```

## 3. Условные операторы (If/Else)
**Задание:** Напишите программу, которая спрашивает у пользователя его возраст.
- Если ему меньше 18, выведите "Вы несовершеннолетний."
- Если от 18 до 64, выведите "Вы взрослый."
- Если 65 или больше, выведите "Вы пенсионер."

```python
age = int(input("Сколько вам лет? "))

if age < 18:
    print("Вы несовершеннолетний.")
elif age <= 64:
    print("Вы взрослый.")
else:
    print("Вы пенсионер.")
```

## 4. Циклы for
**Задание:** Напишите цикл `for`, который выводит все четные числа от 1 до 20.

*Подсказка:* Вы можете использовать функцию `range()` или оператор остатка от деления (`%`).

```python
# Способ 1: использование шага в range()
for i in range(2, 21, 2):
    print(i)

print("-----------")

# Способ 2: использование остатка от деления
for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

## 5. Циклы while
**Задание:** Напишите цикл `while`, который работает как таймер обратного отсчета. Начните с 10 и выводите каждое число до 1. После вывода 1 напечатайте "Пуск!".

```python
count = 10
while count > 0:
    print(count)
    count -= 1
print("Пуск!")
```

## 6. Списки (Lists)
**Задание:** Создайте список, содержащий пять ваших любимых фруктов.
1. Выведите весь список.
2. Выведите первый и последний фрукт в списке, используя индексацию.
3. Добавьте новый фрукт в конец списка, используя метод `.append()`.
4. Удалите один фрукт из списка.

```python
fruits = ["Яблоко", "Банан", "Апельсин", "Груша", "Киви"]

# 1. Выведите весь список.
print("Весь список:", fruits)

# 2. Выведите первый и последний фрукт в списке, используя индексацию.
print("Первый фрукт:", fruits[0])
print("Последний фрукт:", fruits[-1])

# 3. Добавьте новый фрукт в конец списка, используя метод .append().
fruits.append("Манго")
print("После добавления:", fruits)

# 4. Удалите один фрукт из списка.
fruits.remove("Банан")
print("После удаления:", fruits)
```

## 7. Словари (Dictionaries)
**Задание:** Создайте словарь, в котором хранится информация о питомце (например, `"name": "Барон"`, `"animal": "Собака"`, `"age": 3`).
1. Выведите имя питомца.
2. Обновите возраст питомца, сделав его на год старше.
3. Добавьте новую пару ключ-значение для цвета (`"color"`).

```python
pet = {
    "name": "Барон",
    "animal": "Собака",
    "age": 3
}

# 1. Выведите имя питомца.
print("Имя питомца:", pet["name"])

# 2. Обновите возраст питомца, сделав его на год старше.
pet["age"] += 1
print("Новый возраст:", pet["age"])

# 3. Добавьте новую пару ключ-значение для цвета ("color").
pet["color"] = "Коричневый"
print("Обновленный словарь:", pet)
```

## 9. Мини-проект: Угадай число
**Задание:**
1. Используйте модуль `random`, чтобы загадать секретное число от 1 до 10 (`import random`, `secret_number = random.randint(1, 10)`).
2. Попросите пользователя угадать число.
3. Используйте цикл `while`, чтобы продолжать спрашивать, пока пользователь не угадает правильно.
4. Если его догадка слишком велика, выведите "Слишком много!". Если слишком мала - "Слишком мало!".
5. Когда он угадает правильно, выведите поздравительное сообщение.

```python
import random

secret_number = random.randint(1, 10)
guess = None

print("Я загадал число от 1 до 10. Попробуйте угадать!")

while guess != secret_number:
    # input() возвращает строку, поэтому нужно перевести её в целое число (int)
    guess = int(input("Ваша догадка: "))
    
    if guess > secret_number:
        print("Слишком много!")
    elif guess < secret_number:
        print("Слишком мало!")
    else:
        print("Поздравляю! Вы угадали число.")
```
